# Q-1
## Descriptive Statistics 

### it is the branch of statistics that focuses on summarizing, organizing, and presenting raw data in a meaningful way.

### It does not make predictions or infer conclusions beyond the data—it only describes what’s in front of you.


## Inferential Statistics 

### it is the branch of statistics that allows us to make predictions or draw conclusions about a larger group (called a population) by analyzing a smaller sample of data taken from it.

# Difference Between Descriptive and Inferential Statistics

| Descriptive Statistics | Inferential Statistics |
|-------------------------|------------------------|
| Summarizes data | Draws conclusions from sample data |
| Describes existing data | Predicts population characteristics |
| Uses graphs and summary measures | Uses probability and hypothesis testing |
| No predictions | Makes predictions and decisions |

# Q-2

```python
import pandas as pd

data = {
    "Name": ["Ahmed", "Ali", "Sara", "John", "Aisha"],
    "Age": [23, 26, 24, 29, 27],
    "Department": ["Sales", "Marketing", "HR", "Sales", "IT"],
    "Salary": [45000, 52000, 38000, 52000, 70000]
}

df= pd.DataFrame(data)
df
```

```python
df.describe()
```

### Explanation of describe()

- **count** → Number of non-missing values.
- **mean** → Average value.
- **std** → Standard deviation (spread of data).
- **min** → Smallest value.
- **25%** → First Quartile (25% of data lies below this value).
- **50%** → Median of the dataset.
- **75%** → Third Quartile (75% of data lies below this value).
- **max** → Largest value.

# Q-3

```python
print("Mean using Pandas:")
print(df["Salary"].mean())
```

```python
import numpy as np

print("Mean using NumPy:")
print(np.mean(df["Salary"]))
```

```python
print("Median using Pandas:")
print(df["Salary"].median())
```

```python
print("Median using NumPy:")
print(np.median(df["Salary"]))
```

```python
print("Mode:")
print(df["Salary"].mode()[0])
```

```python
import matplotlib.pyplot as plt
```

```python
import matplotlib.pyplot as plt

mean=df['Salary'].mean()
median=df['Salary'].median()
mode=df['Salary'].mode()[0]

values=[mean,median,mode]
labels=['mean','median','mode']

plt.bar(labels,values)
plt.ylabel('Salary')
plt.title('comparision of mean , median and mode')
plt.show()
```
