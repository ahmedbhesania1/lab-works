```python
from scipy import stats
import numpy as np
pop_mean=500
pop_std=12
n=36
samp_mean=496
alpha=0.05
```

```python
z_score=(samp_mean-pop_mean)/(pop_std/np.sqrt(n))
print(f"z-score = {z_score:.2f}")
```

```python
critical_val=stats.norm.ppf(1-(alpha/2))
p_val=2*(1-stats.norm.cdf(abs(z_score)))

```

```python
if z_score>critical_val or z_score<-critical_val:
    print("Reject the null hypothesis")
else:
    print("Fail to reject the null hypothesis")
```

```python
pop_mean=30
samp_std=np.std([28, 31, 35, 29, 27, 32, 30, 34, 26, 33, 31, 29])
n=12
samp_mean=np.mean([28, 31, 35, 29, 27, 32, 30, 34, 26, 33, 31, 29])
alpha=0.05

#t-test
t_score=(samp_mean-pop_mean)/(samp_std/np.sqrt(n))
critical_val=stats.t.ppf(1-(alpha/2),df=n-1)
print(f"critical-value = {critical_val:.2f}")
print(f"t-score = {t_score:.2f}")
if t_score>critical_val or t_score<-critical_val:
    print("Reject the null hypothesis")
else:
    print("Fail to reject the null hypothesis")
```
