# Q-1

```python


math = [75, 80, 65, 90, 85, 70, 60, 95, 88, 78]
physics = [72, 78, 68, 88, 84, 69, 62, 92, 85, 76]
import pandas as pd
df = pd.DataFrame({"Math": math,"physics":physics})
print(df)

```

```python
print(df.cov())
```

important value is 109.06 which is positive means increasing maths marks will increase physics marks

```python
print(df.corr(method='pearson'))
```


 strong correlation which is 0.99 means increasing maths marks will increase physics marks 

```python
import numpy as np
np.random.shuffle(math)
math
```

```python

df=pd.DataFrame({"Math": math,"physics":physics})
df.corr(method='spearman')
```

the rank relation become weak after shuffling

```python
data=np.random.uniform(0,10,100000)
data
```

```python
import matplotlib.pyplot as plt 
plt.hist(data,bins=50)
plt.title("Uniform Distribution")
plt.show()
```

```python
data=np.random.binomial(10,0.5,1000)
```

```python
plt.hist(data,bins=20)
plt.title("Binomial Distribution")
plt.show()
```

```python
customers=np.random.poisson(5,1000)
plt.hist(customers,bins=20)
plt.title("Poisson Distribution")
plt.show()
```

# Q-3

```python
dt=np.random.exponential(2,10000)
dt
plt.hist(dt)
plt.show()
```

```python
sample_mean=[]

for i in range(1000):
    sample=np.random.choice(dt,30)
    sample_mean.append(np.mean(sample))
```

```python
plt.hist(sample_mean,bins=30)
plt.title("Sample Means (n=30)")
plt.show()
```
