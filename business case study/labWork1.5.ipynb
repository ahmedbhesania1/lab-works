```python
c_interval=stats.t.interval(confidence=0.95,
df=len(score)-1,loc=samp_mean,
scale=samp_std/np.sqrt(len(score)))

print(f'confidence interval:{c_interval}')
print(f"we are 95% confident that the true mean falls within {c_interval}")
```

# Q-2

```python
pop_mean=70
pop_std=10
sample=np.random.normal(70,10,100)
samp_mean=np.mean(sample)
alpha=0.05
print(f'sample mean: {samp_mean:.2f}')
```

```python
z_score=(samp_mean-pop_mean)/(pop_std/np.sqrt(len(sample)))
print(f'z_score:{z_score:.2f}')
critical_value=stats.norm.ppf(1-((alpha)/2))
print(f'critical value:{critical_value:.2f}')
p_value=2*(1-stats.norm.cdf(abs(z_score)))
print(f'p-value:{p_value:.2f}')

```

### null hypothesis is that mean is 70
### alternate hypothesis is that mean is not 70
### two tailed test so alpha=0.05

```python
if z_score>critical_value or z_score<-critical_value:
    print("reject null hypothesis")
else:
    print("fail to reject null hypothesis")
```

```python

if p_value>alpha:
    print("fail to reject null hypothesis")
else:
    print("reject null hypothesis")
```

```python
pop_mean=70
alpha=0.05
```

```python
t_stat,p_value=stats.ttest_1samp(samp_weight,pop_mean)
print(f't-statistic: {t_stat:.2f}')
print(f'p-value: {p_value:.4f}')
```

### null hypothesis is that the population mean is equal to 70
### alternative hypothesis is that the population mean is not equal to 70 so two tailed 

```python
if p_value>alpha:
    print("fail to reject null hypothesis")
else:
    print("reject null hypothesis")
```
