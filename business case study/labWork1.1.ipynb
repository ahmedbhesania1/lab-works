```python
#Q-3

import pandas as pd

data={
    "name":["ahmed","ali","mehmod","salah","sarah"],
    "age":[25, 30, 35, 28, 32],
    "salary":[50000, 55000, 60000, 52000, 65000],
    "department":["sales","marketing","hr","sales","it"]
}

df=pd.DataFrame(data)
print(df)

```

```python
print(df.dtypes)
```

```python
print(df.select_dtypes(include="number"))
```

```python
print(df.select_dtypes(include="object"))
```

```python
#Q-4

import matplotlib.pyplot as plt

plt.hist(df["age"])
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")

plt.show()


```

```python
department_count = df["department"].value_counts()

department_count.plot(kind="bar")

plt.title("Department Count")
plt.xlabel("Department")
plt.ylabel("Count")

plt.show()
```

```python
department_count.plot(
    kind="pie",
    autopct="%1.1f%%"
    
)

plt.title("Department Percentage")

plt.ylabel("")

plt.show()
```

```python
plt.boxplot(df["salary"])

plt.title("Salary Distribution")

plt.show()
```

```python
#Q-5

import numpy as np
population = np.random.randint(40, 101, 1000)
sample=np.random.choice(population,100)
```

```python
population_mean = np.mean(population)

population_std = np.std(population)

print("Population Mean:", population_mean)

print("Population Standard Deviation:", round(population_std,2))
```

```python
sample_mean = np.mean(sample)

sample_std = np.std(sample, ddof=1)

print("Sample Mean:", sample_mean)

print("Sample Standard Deviation:", round(sample_std,2))
```

# Population vs Sample

-  includes all observations in the dataset, while a sample is a smaller subset selected from the population. 


- Samples are used because they are faster, less expensive, and easier to analyze while still providing reliable estimates of the population.
