# Q-2

```python
grp_1=np.random.normal(70,10,50)
grp_2=np.random.normal(75,10,50)

from scipy import stats
t_stats,p_value=stats.ttest_ind(grp_1,grp_2)
print("t-statistics:",t_stats)
print("p-value:",p_value)
```

# Interpretation of p-value

If p-value < 0.05:
- Reject the Null Hypothesis.
- There is a statistically significant difference between the two groups.

If p-value > 0.05:
- Fail to reject the Null Hypothesis.
- There is no statistically significant difference between the groups.

# Q-3

```python
import pandas as pd

df = pd.DataFrame({
    "Before":[45,50,55,60,52,48,58,62,57,53],
    "After":[55,60,65,68,61,57,66,72,64,60]
})

df
```

```python
from scipy.stats import ttest_rel

t_stat, p_value = ttest_rel(df["Before"], df["After"])

print("t-statistic =", t_stat)
print("p-value =", p_value)
```

```python
from scipy import stats

mean = df["After"].mean()

sem = stats.sem(df["After"])

confidence = stats.t.interval(
    confidence=0.95,
    df=len(df["After"])-1,
    loc=mean,
    scale=sem
)


print("95% Confidence Interval:", confidence)
```

```python
alpha = 0.05

if p_value < alpha:
    print("Reject the Null Hypothesis")
else:
    print("Fail to Reject the Null Hypothesis")
```

# Type I Error (False Positive)

A Type I Error occurs when we reject the null hypothesis even though it is true.

Example:
A medical test says a healthy person has a disease.

---

# Type II Error (False Negative)

A Type II Error occurs when we fail to reject the null hypothesis even though it is false.

Example:
A medical test says a diseased person is healthy.
